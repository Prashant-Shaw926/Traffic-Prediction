# Dataset inspection

Load the **unmodified** Kaggle file at `data/raw/traffic.csv` and record shape, columns, missingness, junctions, and time frequency.

This notebook does not clean data or train models. If the CSV is missing, `load_raw_traffic()` prints exact download steps and stops.

```python
import sys
from pathlib import Path

from IPython.display import display

ROOT = Path.cwd().resolve()
if not (ROOT / "ml").is_dir():
    ROOT = ROOT.parents[1]
sys.path.insert(0, str(ROOT))

from ml.data.load_data import RAW_CSV_PATH, load_raw_traffic
from ml.data.inspect_data import build_inspection_report

print("Repository root:", ROOT)
print("Raw CSV path:", RAW_CSV_PATH)

df = load_raw_traffic()
print("Loaded rows:", len(df))
```

## Shape, columns, and dtypes

Do not assume paper field names. The next cells report what is actually in the file.

```python
print("shape:", df.shape)
print("columns:", list(df.columns))
print()
print(df.dtypes)
```

## Missing values and duplicate rows

```python
print(df.isna().sum())
print("total missing cells:", int(df.isna().sum().sum()))
print("full-row duplicates:", int(df.duplicated().sum()))
```
